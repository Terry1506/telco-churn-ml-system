**Project Summary**

This project focused on predicting customer churn using the Telco Customer Churn dataset, with the goal of identifying customers at risk of leaving and understanding the drivers behind churn behavior. The workflow followed a structured machine learning pipeline, including data understanding, cleaning and preprocessing, exploratory data analysis, baseline modeling, and comparison of multiple classification models. Logistic Regression, Decision Tree, and Random Forest models were evaluated using ROC-AUC as the primary metric. After hyperparameter tuning, Logistic Regression emerged as the final selected model due to its strong performance and high interpretability. The resulting model provides both predictive power and actionable insights for business decision-making.

**Final Model Recap**

The final selected model for deployment is Logistic Regression, achieving a ROC-AUC score of 0.8459, the highest among all evaluated models.

While the Random Forest model achieved a comparable ROC-AUC, Logistic Regression was preferred due to its transparency and ease of interpretation. In a business context such as customer retention, understanding why a customer is likely to churn is as important as accurately predicting churn itself. Logistic Regression allows stakeholders to directly interpret feature coefficients and align model outputs with customer behavior and operational strategies.

**High-Level Business Insights**

Analysis of the model and data reveals clear behavioral patterns associated with customer churn. Customers early in their lifecycle are significantly more likely to churn, indicating that the onboarding and early experience phases are critical. Flexible, month-to-month billing behaviors are also associated with higher churn risk, suggesting lower commitment and weaker customer attachment. In contrast, long-term contracts serve as strong retention anchors, substantially reducing churn likelihood. These patterns highlight the importance of early engagement, pricing clarity, and contract-based loyalty incentives.

**Limitations of the Current Approach**

Despite its strong performance, the current modeling approach has several limitations. Logistic Regression assumes linear and additive relationships between features and churn probability, which may oversimplify complex customer behaviors. Interaction effects and nonlinear patterns are not explicitly captured. Additionally, the dataset represents a single organization and time period, which may limit generalizability to other contexts. Temporal dynamics, such as how customer behavior changes over time, were also not modeled. These factors should be considered when interpreting results and planning future improvements.

**Recommendations for Improvement**

Future iterations of this project could benefit from exploring more expressive models such as gradient boosting algorithms to capture nonlinear relationships. Incorporating time-based features or customer interaction histories could improve predictive accuracy and early churn detection. Evaluating models using cost-sensitive metrics would better align performance with business objectives, such as minimizing retention campaign costs. Finally, validating the model on newer customer cohorts would help assess long-term stability and robustness.

**Deployment Readiness**

The final model has been saved and versioned, with a clearly defined feature set consistent with preprocessing steps. Its lightweight nature makes it suitable for both batch prediction and real-time API deployment. The interpretability of Logistic Regression supports trust and adoption by non-technical stakeholders, enabling seamless integration into decision-support systems for customer retention.

**Final Takeaway**

Customer churn is driven less by demographics and more by behavioral commitment and early customer experience. By combining strong predictive performance with interpretability, the selected model provides actionable insights that can guide targeted retention strategies. With further refinement and validation, this approach can meaningfully support data-driven customer lifecycle management and reduce revenue loss due to churn.
