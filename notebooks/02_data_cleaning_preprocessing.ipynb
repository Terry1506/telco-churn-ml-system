Import Libraries

```python
import pandas as pd
import numpy as np
```

Load Raw Data

```python
# load raw dataset
df=pd.read_csv('../data/raw/Telco-customer-churn.csv')

#peek at the data
df.head()
```

 Check Missing Values

```python
# Check missing values
missing = df.isnull().sum()
print("Missing values per column:\n", missing)
```

Handle Missing Values

```python
# Convert TotalCharges to numeric (coerce errors to NaN)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Fill missing values with median safely
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())

```

Check Duplicates

```python
# Check for duplicate rows
duplicates = df.duplicated().sum()
print(f"Duplicate rows: {duplicates}")

# Optionally, drop duplicates
df.drop_duplicates(inplace=True)
```

Convert Data Types

```python
# Some numeric columns may be object type
numeric_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric)
df[numeric_cols].dtypes
```

Encode Categorical Variables

```python
# Identify categorical columns
cat_cols = df.select_dtypes(include='object').columns.tolist()

# Remove ID and target if present
if 'customerID' in cat_cols:
    cat_cols.remove('customerID')
if 'Churn' in cat_cols:
    cat_cols.remove('Churn')

# One-hot encode the remaining categorical columns
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)

# Encode target variable
df['Churn'] = df['Churn'].map({'Yes':1, 'No':0})

# Quick check
df.head()


```

Save Cleaned Data

```python
# Save processed dataset
df.to_csv('../data/processed/Telco-Customer-Churn-processed.csv', index=False)
```

Final Observations

Data Cleaning & Preprocessing Observations

- Missing values in 'TotalCharges' handled by median imputation
- Duplicate rows removed
- All categorical variables encoded as numeric
- Target variable 'Churn' encoded: 1 = Yes, 0 = No
- Dataset ready for modeling: clean, numeric, and free of nulls
